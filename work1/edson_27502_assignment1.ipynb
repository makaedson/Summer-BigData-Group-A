## Task 2: Advanced Calculator

## Task 4: String Processing Pipeline

```python
raw_data = " MUGISHA Jean Pierre | 1998 | Kigali, Rwanda | Computer Science | 3.72 "

sanitized_data = raw_data.strip()
fields = [field.strip() for field in sanitized_data.split('|')]

raw_name = fields[0]
name_title = raw_name.title()
birth_year = int(fields[1])
current_age = 2026 - birth_year

location_field = fields[2]
city, country = [loc.strip() for loc in location_field.split(',')]

major_field = fields[3]
has_science = "Science" in major_field
gpa_value = float(fields[4])

name_parts = raw_name.split()
last_name = name_parts[0].title()
first_name = " ".join(name_parts[1:]).title()
full_name = f"{first_name} {last_name}"

print(f"="*40)
print(f"EXTRACTED DATA SUMMARY")
print(f"="*40)
print(f"Full Name Variable : {full_name}")
print(f"Formatted Name     : {name_title}")
print(f"Calculated Age     : {current_age} years old")
print(f"Extracted City     : {city}")
print(f"Extracted Country  : {country}")
print(f"Major Bounding     : {major_field}")
print(f"Contains 'Science' : {has_science}")
print(f"Metric Score GPA   : {gpa_value:.2f}")
print(f"="*40)
```

## Task 5: Financial Calculator

```python
principal = int(input("Enter principal amount in RWF: "))
rate = float(input("Enter annual interest rate (e.g., 12 for 12%): "))
years = int(input("Enter total matrix years: "))
monthly_expense = float(input("Enter monthly expense amount in RWF: "))

simple_interest = (principal * rate * years) / 100
total_simple = principal + simple_interest

total_compound = principal * ((1 + (rate / 100)) ** years)
compound_interest = total_compound - principal

interest_difference = total_compound - total_simple
total_expenses = monthly_expense * 12 * years
net_savings = total_compound - total_expenses

print(f"\n" + "="*50)
print(f"FINANCIAL SUMMARY ANALYSIS REPORT")
print(f"="*50)
print(f"Initial Principal Base      : {principal:,.2f} RWF")
print(f"Simple Interest Yield       : {simple_interest:,.2f} RWF")
print(f"Total Simple Balance        : {total_simple:,.2f} RWF")
print(f"Compound Interest Yield     : {compound_interest:,.2f} RWF")
print(f"Total Compound Balance      : {total_compound:,.2f} RWF")
print(f"Delta Interest Margin       : {interest_difference:,.2f} RWF")
print(f"Cumulative Matrix Expenses  : {total_expenses:,.2f} RWF")
print(f"Net Period Savings Metric   : {net_savings:,.2f} RWF")
print(f"="*50)
```
