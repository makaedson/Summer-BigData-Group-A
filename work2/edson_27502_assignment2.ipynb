# Task 1: Smart Grade Classifier

```python
def get_grade(score):
    if 90 <= score <= 100:
        return 'A', 'Distinction'
    elif 80 <= score < 90:
        return 'B', 'Merit'
    elif 70 <= score < 80:
        return 'C', 'Credit'
    elif 60 <= score < 70:
        return 'D', 'Pass'
    elif score < 60:
        return 'F', 'Fail'
    else:
        return 'A', 'Distinction'

students = [
    ("Mugabo Edson", 95),
    ("Alice Uwase", 88),
    ("Bob Mugisha", 72),
    ("Carol Ingabire", 91),
    ("David Habimana", 45),
    ("Eve Mutoni", 67),
    ("Frank Nkurunz", 78),
    ("Grace Umutoni", 82),
    ("Henri Gasabo", 55),
    ("Ian Kagame", 61)
]

print(f"{'Name':<25} | {'Score':<5} | {'Grade':<5} | {'Remark':<15}")
print("-" * 60)

total_students = len(students)
grade_counts = {'A': 0, 'B': 0, 'C': 0, 'D': 0, 'F': 0}
highest_score = float('-inf')
lowest_score = float('inf')
total_score = 0
passed_count = 0

for name, score in students:
    grade, remark = get_grade(score)
    print(f"{name:<25} | {score:<5} | {grade:<5} | {remark:<15}")
    
    grade_counts[grade] += 1
    total_score += score
    if score > highest_score:
        highest_score = score
    if score < lowest_score:
        lowest_score = score
    if grade != 'F':
        passed_count += 1

class_average = total_score / total_students
pass_rate = (passed_count / total_students) * 100

print("\n" + "="*40)
print("CLASS SUMMARY STATISTICS")
print("="*40)
print(f"Total Students processed : {total_students}")
print(f"Grade Categories count   : {dict(grade_counts)}")
print(f"Highest Score in Class   : {highest_score}")
print(f"Lowest Score in Class    : {lowest_score}")
print(f"Class Average Score      : {class_average:.2f}")
print(f"Class Pass Rate          : {pass_rate:.2f}%")
```

# Task 2: Number Analysis Tool

```python
def full_number_analysis(numbers):
    count = 0
    total_sum = 0
    minimum = None
    maximum = None
    positive_count = 0
    negative_count = 0
    zero_count = 0
    even_count = 0
    odd_count = 0
    
    for num in numbers:
        count += 1
        total_sum += num
        
        if minimum is None or num < minimum:
            minimum = num
        if maximum is None or num > maximum:
            maximum = num
            
        if num > 0:
            positive_count += 1
        elif num < 0:
            negative_count += 1
        else:
            zero_count += 1
            
        if num % 2 == 0:
            even_count += 1
        else:
            odd_count += 1
            
    average = total_sum / count if count > 0 else 0
    
    above_average_count = 0
    for num in numbers:
        if num > average:
            above_average_count += 1
            
    return (
        count,
        total_sum,
        average,
        minimum,
        maximum,
        positive_count,
        negative_count,
        zero_count,
        even_count,
        odd_count,
        above_average_count
    )

number_list = [12, -7, 0, 15, -3, 8, 22, -14, 0, 5, 11, -9, 18, -4, 25]

analysis_results = full_number_analysis(number_list)

(cnt, tot_sum, avg, min_val, max_val, pos_cnt, neg_cnt, zero_cnt, even_cnt, odd_cnt, above_avg_cnt) = analysis_results

print("==============================")
print("   NUMBER ANALYSIS RESULTS    ")
print("==============================")
print(f"Total Count            : {cnt}")
print(f"Sum of All Numbers     : {tot_sum}")
print(f"Average of Numbers     : {avg:.2f}")
print(f"Minimum Value          : {min_val}")
print(f"Maximum Value          : {max_val}")
print(f"Positive Numbers Count : {pos_cnt}")
print(f"Negative Numbers Count : {neg_cnt}")
print(f"Zeros Count            : {zero_cnt}")
print(f"Even Numbers Count     : {even_cnt}")
print(f"Odd Numbers Count      : {odd_cnt}")
print(f"Numbers Above Average  : {above_avg_cnt}")
print("------------------------------")

positive_numbers = [num for num in number_list if num > 0]
even_numbers = [num for num in number_list if num % 2 == 0]
above_average_numbers = [num for num in number_list if num > avg]

print(f"Original List                  : {number_list}")
print(f"Filtered (Only Positives)      : {positive_numbers}")
print(f"Filtered (Only Evens)          : {even_numbers}")
print(f"Filtered (Greater than Average): {above_average_numbers}")
```

# Task 3: Pattern Generator

```python
def print_section_divider(pattern_name, n):
    print(f"\n--- {pattern_name} (n = {n}) ---")

def print_right_triangle(n):
    for i in range(1, n + 1):
        for j in range(i):
            print("*", end=" ")
        print()

def print_inverted_triangle(n):
    for i in range(n, 0, -1):
        for j in range(i):
            print("*", end=" ")
        print()

def print_pyramid(n):
    for i in range(1, n + 1):
        print(" " * (n - i), end="")
        for j in range(i):
            print("*", end=" ")
        print()

def print_hollow_rectangle(n):
    height = n // 2
    if height < 1:
        height = 1
    
    for i in range(height):
        if i == 0 or i == height - 1:
            print("* " * n)
        else:
            print("* " + "  " * (n - 2) + "*")

def print_number_triangle(n):
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            print(j, end=" ")
        print()

for size in [5, 7]:
    print_section_divider("Right-Angled Triangle", size)
    print_right_triangle(size)
    
    print_section_divider("Inverted Triangle", size)
    print_inverted_triangle(size)
    
    print_section_divider("Centered Pyramid", size)
    print_pyramid(size)
    
    print_section_divider("Hollow Rectangle", size)
    print_hollow_rectangle(size)
    
    print_section_divider("Number Triangle", size)
    print_number_triangle(size)
```

# Task 4: Mini Banking System

```python
def deposit(balance, amount):
    if amount <= 0:
        print("Error: Deposit amount must be positive.")
        return balance
    return balance + amount

def withdraw(balance, amount):
    if amount <= 0:
        print("Error: Withdrawal amount must be positive.")
        return balance
    if amount > balance:
        print(f"Error: Insufficient funds. Current balance is RWF {balance:,.2f}.")
        return balance
    return balance - amount

def get_statement(transactions):
    print("\n" + "="*50)
    print("                    MINI BANK STATEMENT")
    print("="*50)
    print(f"{'No.':<4} | {'Type':<12} | {'Amount (RWF)':<15} | {'Balance (RWF)':<15}")
    print("-" * 50)
    for idx, txn in enumerate(transactions, 1):
        print(f"{idx:<4} | {txn['type']:<12} | {txn['amount']:>15,.2f} | {txn['balance_after']:>15,.2f}")
    print("="*50 + "\n")

def run_banking_system():
    balance = 50000.0
    transactions = []
    
    print("Welcome to the Mini Banking System!")
    
    mock_inputs = ['1', '10000', '2', '5000', '3', '4']
    
    def get_input(prompt):
        if mock_inputs:
            val = mock_inputs.pop(0)
            print(f"{prompt}{val} (Simulated)")
            return val
        return input(prompt)
    
    while True:
        print("\n--- MENU ---")
        print("1. Deposit")
        print("2. Withdraw")
        print("3. View Statement")
        print("4. Exit")
        
        choice = get_input("Enter your choice (1-4): ").strip()
        
        if choice == '1':
            try:
                amount_str = get_input("Enter amount to deposit: ").strip()
                amount = float(amount_str)
                new_balance = deposit(balance, amount)
                
                if new_balance > balance:
                    balance = new_balance
                    transactions.append({
                        'type': 'Deposit',
                        'amount': amount,
                        'balance_after': balance
                    })
                    print(f"Success: Deposited RWF {amount:,.2f}. Current balance: RWF {balance:,.2f}")
            except ValueError:
                print("Error: Invalid input. Please enter a valid numeric amount.")
                
        elif choice == '2':
            try:
                amount_str = get_input("Enter amount to withdraw: ").strip()
                amount = float(amount_str)
                new_balance = withdraw(balance, amount)
                
                if new_balance < balance:
                    balance = new_balance
                    transactions.append({
                        'type': 'Withdrawal',
                        'amount': amount,
                        'balance_after': balance
                    })
                    print(f"Success: Withdrew RWF {amount:,.2f}. Current balance: RWF {balance:,.2f}")
            except ValueError:
                print("Error: Invalid input. Please enter a valid numeric amount.")
                
        elif choice == '3':
            if not transactions:
                print("No transactions recorded yet.")
            else:
                get_statement(transactions)
                
        elif choice == '4':
            print("\nThank you for using the Mini Banking System.")
            print(f"Final Balance: RWF {balance:,.2f}")
            print(f"Total Transactions Recorded: {len(transactions)}")
            break
            
        else:
            print("Error: Invalid choice. Please choose an option from 1 to 4.")

run_banking_system()
```

# Task 5: Data Processing Pipeline

```python
raw_data = [
    " Alice Uwase      | 22 | F | Kigali    | 88.5 | Computer Science ",
    " Bob Mugisha      | 19 | M | Butare    | 72.0 | Mathematics      ",
    " Carol Ingabire   | 24 | F | Gisenyi   | 95.5 | Data Science     ",
    " David Habimana   | 21 | M | Kigali    | 41.0 | Computer Science ",
    " Eve Mutoni       | 20 | F | Rwamagana | 65.5 | Mathematics      ",
    " Frank Nkurunz    | 23 | M | Butare    | 78.0 | Data Science     ",
    " Grace Umutoni    | 18 | F | Kigali    | 91.0 | Computer Science ",
    " Henri Gasabo     | 22 | M | Musanze   | 55.5 | Mathematics      "
]

def parse_record(raw_string):
    parts = raw_string.split('|')
    return {
        'name': parts[0].strip(),
        'age': int(parts[1].strip()),
        'gender': parts[2].strip(),
        'city': parts[3].strip(),
        'score': float(parts[4].strip()),
        'department': parts[5].strip()
    }

def clean_dataset(raw_list):
    cleaned_list = []
    for raw_str in raw_list:
        cleaned_list.append(parse_record(raw_str))
    return cleaned_list

def filter_by_department(data, dept):
    filtered = []
    for record in data:
        if record['department'].lower() == dept.lower():
            filtered.append(record)
    return filtered

def top_performers(data, n):
    sorted_data = sorted(data, key=lambda x: x['score'], reverse=True)
    return sorted_data[:n]

def get_average_score_per_department(data):
    dept_scores = {}
    for record in data:
        dept = record['department']
        score = record['score']
        if dept not in dept_scores:
            dept_scores[dept] = []
        dept_scores[dept].append(score)
        
    dept_averages = {}
    for dept, scores in dept_scores.items():
        dept_averages[dept] = sum(scores) / len(scores)
    return dept_averages

def get_female_students_from_kigali(data):
    filtered = []
    for record in data:
        if record['gender'].upper() == 'F' and record['city'].lower() == 'kigali':
            filtered.append(record)
    return filtered

cleaned_data = clean_dataset(raw_data)

print("=" * 70)
print("                  DATA PROCESSING PIPELINE REPORT")
print("=" * 70)

print("\n1. ALL CLEANED STUDENT RECORDS:")
print(f"{'Name':<18} | {'Age':<3} | {'Gender':<6} | {'City':<10} | {'Score':<5} | {'Department':<16}")
print("-" * 70)
for s in cleaned_data:
    print(f"{s['name']:<18} | {s['age']:<3} | {s['gender']:<6} | {s['city']:<10} | {s['score']:<5.1f} | {s['department']:<16}")

print("\n2. AVERAGE SCORE PER DEPARTMENT:")
dept_averages = get_average_score_per_department(cleaned_data)
for dept, avg_score in dept_averages.items():
    print(f"   * {dept:<20}: {avg_score:.2f}")

print("\n3. TOP 3 PERFORMERS OVERALL:")
top_3 = top_performers(cleaned_data, 3)
for rank, s in enumerate(top_3, 1):
    print(f"   Rank {rank}: {s['name']} - Score: {s['score']} ({s['department']})")

print("\n4. FEMALE STUDENTS FROM KIGALI:")
females_kigali = get_female_students_from_kigali(cleaned_data)
if females_kigali:
    for s in females_kigali:
        print(f"   * {s['name']} - Age: {s['age']} - Score: {s['score']} ({s['department']})")
else:
    print("   No female students from Kigali found.")
print("=" * 70)
```
